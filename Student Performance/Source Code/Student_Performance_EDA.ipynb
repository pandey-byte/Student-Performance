## Question 1: What are the average first-period (G1), second-period (G2) and final (G3) grades?
*(Calculated using NumPy arrays)*

```python
# Convert the grade columns to NumPy arrays
g1 = df['G1'].to_numpy()
g2 = df['G2'].to_numpy()
g3 = df['G3'].to_numpy()

print("Average G1 (first period) :", round(np.mean(g1), 2))
print("Average G2 (second period):", round(np.mean(g2), 2))
print("Average G3 (final)        :", round(np.mean(g3), 2))

# A few more NumPy statistics for context
grades = np.array([g1, g2, g3])          # 2-D array: 3 rows (G1, G2, G3) x 395 columns
summary = pd.DataFrame({
    'Mean':   np.mean(grades, axis=1).round(2),
    'Median': np.median(grades, axis=1),
    'Std Dev': np.std(grades, axis=1).round(2),
    'Min':    np.min(grades, axis=1),
    'Max':    np.max(grades, axis=1)
}, index=['G1', 'G2', 'G3'])
summary
```

**Answer:** The average grades are **G1 = 10.91**, **G2 = 10.71** and **G3 = 10.42** (out of 20). The average drops slightly from period to period, and the spread (std dev) grows from 3.31 to 4.58.

## Question 2: How many students fall into each grade band?
**Chosen grade bands (on final grade G3, scale 0–20):**
- Fail: 0–9
- Pass: 10–12
- Good: 13–15
- Excellent: 16–20

```python
bins   = [-1, 9, 12, 15, 20]
labels = ['Fail (0-9)', 'Pass (10-12)', 'Good (13-15)', 'Excellent (16-20)']

df['grade_band'] = pd.cut(df['G3'], bins=bins, labels=labels)

band_counts = df['grade_band'].value_counts().sort_index()
print(band_counts)

print("\nPercentage of students in each band:")
print((band_counts / len(df) * 100).round(1))
```

**Answer:** Fail = **130**, Pass = **134**, Good = **91**, Excellent = **40** students. About one-third (32.9%) of students failed, and only about 10% reached the Excellent band.

## Question 3: Compare average final grades by gender

```python
df['gender'] = df['sex'].map({'F': 'Female', 'M': 'Male'})

gender_summary = df.groupby('gender')['G3'].agg(['mean', 'count']).round(2)
gender_summary.columns = ['Average G3', 'Number of students']
gender_summary
```

**Answer:** Male students average **10.91** and female students average **9.97** in the final grade – a gap of about 0.94 marks in favour of males (187 male vs 208 female students).

## Question 4: Compare average final grades for different study-time categories
Study time (weekly): **1** = <2 hours, **2** = 2–5 hours, **3** = 5–10 hours, **4** = >10 hours

```python
study_map = {1: '1: <2 hrs', 2: '2: 2-5 hrs', 3: '3: 5-10 hrs', 4: '4: >10 hrs'}
df['study_category'] = df['studytime'].map(study_map)

study_summary = df.groupby('study_category')['G3'].agg(['mean', 'count']).round(2)
study_summary.columns = ['Average G3', 'Number of students']
study_summary
```

**Answer:** Average final grade by study time: <2 hrs = **10.05**, 2–5 hrs = **10.17**, 5–10 hrs = **11.40**, >10 hrs = **11.26**. Students who study more than 5 hours a week score about 1.2 marks higher than those who study less.

## Chart 1: Histogram of final grades (G3)

```python
plt.figure(figsize=(8, 5))
plt.hist(df['G3'], bins=range(0, 22), color='steelblue', edgecolor='black')
plt.axvline(df['G3'].mean(), color='red', linestyle='--', label=f"Mean = {df['G3'].mean():.2f}")
plt.title('Distribution of Final Grades (G3)')
plt.xlabel('Final Grade (G3)')
plt.ylabel('Number of Students')
plt.xticks(range(0, 21, 2))
plt.legend()
plt.show()
```

**Graph answer:** Most students score between 8 and 14, and the mean is about 10.4. There is a noticeable spike at **0** (38 students with a final grade of zero), which stands apart from the rest of the bell-shaped distribution.

## Chart 2: Study-time count chart

```python
study_counts = df['study_category'].value_counts().sort_index()

plt.figure(figsize=(8, 5))
bars = plt.bar(study_counts.index, study_counts.values, color='seagreen', edgecolor='black')
plt.bar_label(bars)                      # show the count on top of each bar
plt.title('Number of Students by Weekly Study Time')
plt.xlabel('Study-time category')
plt.ylabel('Number of Students')
plt.show()
```

**Graph answer:** Most students (198) study 2–5 hours a week, 105 study under 2 hours, 65 study 5–10 hours and only 27 study more than 10 hours. So about 77% of the class studies fewer than 5 hours a week.

## Three Observations

```python
# Extra numbers backing up the observations
print("Students with G3 = 0:", (df['G3'] == 0).sum())
print("Average absences of students with G3 = 0:", df.loc[df['G3'] == 0, 'absences'].mean())
print("Correlation of G3 with G1:", round(df['G3'].corr(df['G1']), 2))
print("Correlation of G3 with G2:", round(df['G3'].corr(df['G2']), 2))
print("Correlation of G3 with absences:", round(df['G3'].corr(df['absences']), 2))
```

1. **Grades are strongly linked across periods, and the average slips slightly.** Mean grades fall from 10.91 (G1) to 10.71 (G2) to 10.42 (G3), and G3 is very strongly correlated with G1 (0.80) and G2 (0.90). A student's early grades are a good indicator of their final result.
2. **More study time goes with higher grades, and males score higher on average.** Students studying 5+ hours a week average about 11.3 versus about 10.1 for those studying under 5 hours. Males average 10.91 vs 9.97 for females. Most students (77%) study less than 5 hours a week.
3. **Absences are not a simple predictor of low grades.** The top absentees (e.g. 75, 56, 54 absences) have mixed results, and the overall correlation between absences and G3 is almost zero (0.03). Also, the 38 students with a final grade of 0 all have 0 absences, which suggests those are students who did not take the final assessment rather than students who were skipping class. The data has no missing values or duplicates.
